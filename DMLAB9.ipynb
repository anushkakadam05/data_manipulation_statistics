```python
coupons = np.random.randint(1,6,1000)
```

```python
coupons_counts = pd.Series(coupons).value_counts().sort_index()
print(coupons_counts)
```

```python
plt.bar(coupons_counts.index,coupons_counts.values)
plt.title("Uniform Distribution of Coupons")
plt.xlabel("Coupon Number")
plt.ylabel("Frequency")
plt.show()
```

```python
probabilities = coupons_counts/coupons_counts.sum()
print(probabilities)
```

```python
uniform_summary = pd.DataFrame({
    "Coupon" : coupons_counts.index,
    "Probability" : probabilities.values
})
uniform_summary
```

```python
review_count = df["Review_Count"]
```

```python
review_groups =pd.cut(review_count,bins = 10)
```

```python
review_distribution = review_groups.value_counts().sort_index()
print(review_distribution)
```

```python
review_distribution.plot(kind = 'bar')
plt.title("Review Count Distribution")
plt.xlabel("Review Count Range")
plt.ylabel("Frequency")
plt.show()
```

```python
review_probabilities = review_distribution/review_distribution.sum()
print(review_probabilities)
```

```python
uniform_summary = pd.DataFrame({
    "Review Count Range" : review_distribution.index.astype(str),
    "Frequency" : review_distribution.values,
    "Probability" : review_probabilities.values
})
uniform_summary
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import binom
```

```python
n =20
```

```python
p = 0.30
```

```python
print("Number of Customers:",n)
print("Number of Probability:",p)
```

```python
prob_5 = binom.pmf(5,n,p)
print(prob_5)
```

```python
prob_6 = binom.pmf(6,n,p)
print(prob_6)
```

```python
prob_10 = binom.pmf(10,n,p)
print(prob_10)
```

```python
x =np.arange(0,n+1)
```

```python
y = binom.pmf(x,n,p)
```

```python
probability_table = pd.DataFrame({
    "Number of Customers" : x,
    "Probability" : y
})
probability_table.head(10)
```

```python
plt.figure(figsize = (10,5))
plt.bar(x,y)
plt.title("Binomial Distribution")
plt.xlabel("Number of Purchases")
plt.ylabel("Probability")
plt.show()
```

```python
most_likely = x[np.argmax(y)]
print("Most Likely Purchases :",most_likely)
```

```python
expected_purchases = n*p
print("Expected Purchases :",expected_purchases)
```

```python
summary = pd.DataFrame({
    "Metric" :[
        "Number of Customers",
        "Purchase Probability",
        "Expected Purchases",
        "Most Likely Purchases"
    ],
    "Value" :[
        n,
        p,
        expected_purchases,
        most_likely
    ]
})
summary
```
