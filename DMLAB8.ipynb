```python
revenue_skew = df["Revenue"].skew()
print(revenue_skew)
```

```python
if revenue_skew > 0:
  print("Positive skew :", revenue_skew)
elif revenue_skew < 0:
  print("Negative skew :", revenue_skew)
else :
  print("Symmetric skew :", revenue_skew)
```

```python
units_skew = df["Units_Sold"].skew()
print(units_skew)
```

```python
import matplotlib.pyplot as plt
```

```python
plt.hist(df["Revenue"], bins = 20)
plt.title("Revenue_Distribution")
plt.xlabel("Revenue")
plt.ylabel("Frequency")
plt.show()
```

```python
plt.hist(df["Units_Sold"], bins = 20)
plt.title("Units_Sold_Distribution")
plt.xlabel("Units_Sold")
plt.ylabel("Frequency")
plt.show()
```

```python
summary = pd.DataFrame({
    "Metric" : ["Revenue","Units_Sold"],
    "Skewness" : [revenue_skew,units_skew]
})
summary
```
